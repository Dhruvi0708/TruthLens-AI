### Objectives

- Load the saved model
- Load the TF-IDF vectorizer
- Predict custom news articles
- Test on unseen examples
- Verify saved model works correctly

```python
import joblib
import pandas as pd
```

```python
model = joblib.load("../models/fake_news_model.pkl")

vectorizer = joblib.load("../models/tfidf_vectorizer.pkl")

print("Model loaded successfully!")
```

```python
def predict_news(text):
    text_vector = vectorizer.transform([text])
    prediction = model.predict(text_vector)[0]

    if prediction == 0:
        return "Fake News"
    else:
        return "Real News"
```

```python
news_df = pd.read_csv("../data/clean_news_dataset.csv")

for i in range(5):
    text = news_df["clean_text"][i]
    actual = "Fake News" if news_df["label"][i] == 0 else "Real News"
    prediction = predict_news(text)

    print(f"Article {i+1}")
    print("Actual     :", actual)
    print("Prediction :", prediction)
    print("-" * 60)
```

```python
fake_news = """
Aliens have landed in New York and declared Earth under Mars government.
"""

print("Prediction:", predict_news(fake_news))
```

```python
real_news = """
NASA has launched a new satellite to monitor climate change across the globe.
"""

print("Prediction:", predict_news(real_news))
```

```python
user_news = input("Enter a news article:\n")

print()

print("Prediction:", predict_news(user_news))
```

```python
new_article = """
Apple Inc. announced strong quarterly earnings, reporting higher-than-expected iPhone sales and increased revenue from its services business. The company said demand remained strong across several international markets.
"""

print(predict_news(new_article))
```
