```python
import pandas as pd
import re
import string
import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
news_df = pd.read_csv("../data/news_dataset.csv")

news_df.head()
```

```python
stop_words = set(stopwords.words("english"))

lemmatizer = WordNetLemmatizer()
```

```python
def clean_text(text):

    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r"http\S+|www\S+", "", text)

    # Remove HTML tags
    text = re.sub(r"<.*?>", "", text)

    # Remove punctuation
    text = text.translate(str.maketrans("", "", string.punctuation))

    # Remove numbers
    text = re.sub(r"\d+", "", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    # Tokenize
    words = text.split()

    # Remove stopwords and lemmatize
    words = [
        lemmatizer.lemmatize(word)
        for word in words
        if word not in stop_words
    ]

    return " ".join(words)
```

```python
news_df["clean_text"] = news_df["text"].apply(clean_text)
```

```python
print("Original Text:\n")
print(news_df["text"][0])

print("\n" + "="*100 + "\n")

print("Cleaned Text:\n")
print(news_df["clean_text"][0])
```

```python
news_df["clean_text"].isnull().sum()
```

```python
# Remove rows where clean_text is missing or empty
news_df = news_df.dropna(subset=["clean_text"])
news_df = news_df[news_df["clean_text"].str.strip() != ""]

# Save cleaned dataset
news_df.to_csv("../data/clean_news_dataset.csv", index=False)

print("Final dataset shape:", news_df.shape)
print("Text preprocessing completed successfully!")
```
